## Tokenization

` ` - 0  
`/` - 1  
`|` - 2  
`\` - 3  
`_` - 4  
`~` - 5  
`&` - 6  

```python
mapping = {
    ' ': '0',
    '/': '1',
    '|': '2',
    '\\': '3',
    '_': '4',
    '~': '5',
    '&': '6',
    '\n': '\n',
}

def tokenize(frame):
    result = ""
    for c in frame:
        result += mapping[c]
    return result
```

```python
f = open("../dataset/raw/tree_0000/frame_0113.txt")
s = f.read()
```

```python
t = tokenize(s)
t
```

```python
from pathlib import Path
import os

os.makedirs("../dataset/tokenized", exist_ok=True)
root = Path("../dataset/raw")

for t_num, tree_dir in enumerate(sorted(root.iterdir())):
    if tree_dir.is_dir():
        os.makedirs(f"../dataset/tokenized/tree_{t_num:04d}", exist_ok=True)
        for f_num, frame in enumerate(sorted(tree_dir.iterdir())):
            if frame.is_file():
                with open(frame) as f2:
                    data = f2.read()
                    t = tokenize(data)
                with open(f"../dataset/tokenized/tree_{t_num:04d}/frame_{f_num:04d}.txt", "w+") as f:
                    f.write(t)
```
